```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score

# 1. قراءة البيانات
df = pd.read_csv(r"D:\diabetes.csv")

# 2. التأكد من عدم وجود قيم مفقودة أو مكررة
print("Missing values:{}".format(df.isnull().any().sum()))
print("Duplicated values:{}".format(df.duplicated().any().sum()))

# 3. تحديد المتغيرات المستقلة (X) والهدف (y) مع إصلاح شكل y ليصبح 1D
X = df[['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']]
y = df['Outcome'].values.ravel()

# 4. معالجة وتطبيع البيانات
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(X)

# 5. تقسيم البيانات إلى تدريب واختبار
x_train, x_test, y_train, y_test = train_test_split(scaled_data, y, test_size=0.2, random_state=42)

# 6. بناء وتدريب النموذج
model = LogisticRegression()
model.fit(x_train, y_train)

# 7. طباعة دقة النموذج
print("Training Accuracy:", model.score(x_train, y_train))
print("Testing Accuracy:", model.score(x_test, y_test))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. قراءة البيانات
df = pd.read_csv(r"D:\diabetes.csv")

# 2. رسم مصفوفة الارتباط (Correlation Heatmap) لمعرفة علاقة الأعمدة ببعضها
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Diabetes Dataset")
plt.show()

# 3. رسم توزيع متغير الهدف (Outcome) لمعرفة عدد الحالات الإيجابية والسلبية
plt.figure(figsize=(6, 4))
sns.countplot(x='Outcome', data=df, palette='Set2')
plt.title("Distribution of Outcome (0: No Diabetes, 1: Diabetes)")
plt.show()

# 4. رسم توزيع الأعمار (Age Distribution)
plt.figure(figsize=(8, 4))
sns.histplot(df['Age'], kde=True, color='purple', bins=20)
plt.title("Age Distribution in Dataset")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# 5. رسم Boxplot لمقارنة نسبة الجلوكوز (Glucose) مع الإصابة (Outcome)
plt.figure(figsize=(6, 4))
sns.boxplot(x='Outcome', y='Glucose', data=df, palette='Pastel1')
plt.title("Glucose Levels vs Outcome")
plt.show()
```
